# Interactive Tour Editor - Visual Interface

This notebook provides an **interactive visual interface** to reorder stops within tours.

## Features

- 🗺️ **Visual Map**: See tours on an interactive map (like the main planning script)
- 📋 **Sortable List**: Reorder stops with up/down buttons
- 🔄 **Real-time Updates**: Map updates as you reorder stops
- ✅ **Easy to Use**: Click buttons to move stops, then apply changes

## How to Use

1. **Setup**: Mount Drive and load modules
2. **Load Data**: Load your VRP results
3. **Select Tour**: Choose a tour from the dropdown
4. **Reorder Stops**: Use ↑ and ↓ buttons to move stops
5. **Apply Changes**: Click 'Apply Changes' to save
6. **Continue**: Use updated tours in your workflow

## Step 3: Load Your Data

You have two options:

**Option A**: Load from a saved JSON file (if you saved VRP results)
**Option B**: Load from variables (if running after main TourPlanning_Script)

Choose one option below.


```python
# @title Option A: Load from JSON File

# ⚠️ UPDATE THIS PATH to your VRP response JSON file
vrp_response_file = '/content/drive/MyDrive/Colab Notebooks/kemmler/2025-12-15/vrp_response.json'  # @param {type:"string"}

# Load VRP response
with open(vrp_response_file, 'r') as f:
    vrp_response_json = json.load(f)

# Load geocoded data (needed for customer names, addresses)
# ⚠️ UPDATE THIS PATH to your geocoded CSV file
geocoded_file = '/content/drive/MyDrive/Colab Notebooks/kemmler/2025-12-15/geocoded_data.csv'  # @param {type:"string"}
df_geocoded = pd.read_csv(geocoded_file)

print("✅ Data loaded successfully!")
print(f"   Tours: {len(vrp_response_json.get('tours', []))}")
print(f"   Geocoded orders: {len(df_geocoded)}")
```

## Step 4: Interactive Tour Editor

The editor will display:
- **Left side**: Interactive map showing the tour route
- **Right side**: List of stops with up/down buttons to reorder

### Instructions:
1. Select a tour from the dropdown
2. View the tour on the map and in the stops list
3. Use ↑ and ↓ buttons to move stops up or down
4. Watch the map update in real-time
5. Click 'Apply Changes' when done
6. Continue with your workflow using the updated `vrp_response_json`


```python
# @title Launch Interactive Editor

# Launch the interactive editor
updated_vrp_response = create_interactive_tour_editor(
    vrp_response_json,
    df_geocoded
)

# Update the main variable
vrp_response_json = updated_vrp_response

print("\n" + "="*80)
print("✅ Editor ready! Use the interface above to reorder stops.")
print("="*80)
```
